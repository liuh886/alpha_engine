# v4.2 SGOV recovery-release experiment

This notebook compares three predeclared ways to release the frozen SGOV defensive sleeve during a market recovery. It does not search weights or thresholds, and it does not change the formal v4.2 state-2 allocation.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('artifacts/evidence/qqqi_qqq_tqqq_v4_2_sgov_recovery_release_v4_4_research')
headline = pd.read_csv(ROOT / 'headline_metrics.csv')
chronological = pd.read_csv(ROOT / 'chronological_metrics.csv')
summary = json.loads((ROOT / 'experiment_summary.json').read_text(encoding='utf-8'))
```

## Headline comparison

```python
columns = ['strategy', 'cagr', 'annual_volatility', 'sharpe', 'sortino', 'max_drawdown', 'calmar', 'turnover_units']
display(headline[columns].sort_values('cagr', ascending=False))
```

## Pre-registered candidate gates

```python
gate_rows = []
for strategy, gate in summary['candidate_gates'].items():
    gate_rows.append({'strategy': strategy, 'authorized': gate['prospective_challenger_authorized'], **gate['metrics']})
display(pd.DataFrame(gate_rows))
```

## Early versus late sample

```python
pivot = chronological.pivot(index='strategy', columns='segment', values='cagr')
display(pivot)
```

## Five major troughs and recovery lag

```python
episode_rows = []
for strategy in ['static_blended', 'qqqi_release_on_state_1', 'tqqq_release_on_precursor', 'staged_qqqi_then_tqqq_release']:
    table = pd.read_csv(ROOT / f'major_drawdown_episodes_{strategy}.csv')
    for _, row in table.iterrows():
        episode_rows.append({
            'strategy': strategy,
            'episode_id': row['episode_id'],
            'protection_pp': row['drawdown_improvement_pp'],
            'recovery_lag_sessions': row['recovery_lag_sessions'],
        })
episodes = pd.DataFrame(episode_rows)
display(episodes)
```

## Precursor-event concentration

```python
for strategy in ['tqqq_release_on_precursor', 'staged_qqqi_then_tqqq_release']:
    events = pd.read_csv(ROOT / f'precursor_events_{strategy}.csv')
    print(strategy, len(events))
    display(events)
```

## Interpretation rule

A candidate must shorten recovery without destroying the blended profile's major-trough protection. Early TQQQ exposure must also work across events and time segments rather than deriving its advantage from one rebound. Passing authorizes prospective monitoring only; v4.2 remains the baseline.
